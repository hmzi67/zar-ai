```python
import json, glob, re, gc, random, unicodedata, numpy as np, torch
from collections import defaultdict
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer, CrossEncoder
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

def find(name):
    hits = glob.glob(f"/kaggle/input/**/{name}", recursive=True)
    assert hits, f"{name} not found under /kaggle/input"
    return hits[0]

chunks = [json.loads(l) for l in open(find("chunks.jsonl"), encoding="utf-8")]
qa = [json.loads(l) for l in open(find("qa.jsonl"), encoding="utf-8")]
qa = [r for r in qa if r["kb_chunk_ids"]]
ids = [c["chunk_id"] for c in chunks]
src = {c["chunk_id"]: c["source_id"] for c in chunks}
N = len(ids)
fact = lambda r: r["notes"].split(";")[0]
facts = sorted({fact(r) for r in qa})
random.Random(42).shuffle(facts)
test_facts = set(facts[len(facts) // 2:])
print(f"KB v3: {N} chunks, {len(qa)} questions, {len(facts)} facts, {len(test_facts)} test facts")

# 1. Roman Urdu to Urdu script (Qwen2.5-14B, same setup as e3_final)
name = "Qwen/Qwen2.5-14B-Instruct"
tokzr = AutoTokenizer.from_pretrained(name)
llm = AutoModelForCausalLM.from_pretrained(name, device_map="auto",
    quantization_config=BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_compute_dtype=torch.float16))
SHOTS = [("gandum ki fasal achhi hai", "گندم کی فصل اچھی ہے"), ("kungi ka hamla ho gaya", "کنگی کا حملہ ہو گیا"),
         ("telay ne patte kha liye", "تیلے نے پتے کھا لیے"), ("khaad aur pani waqt par dein", "کھاد اور پانی وقت پر دیں"),
         ("beej ki sharah barha dein", "بیج کی شرح بڑھا دیں"), ("barani ilaqon mein barish kam hai", "بارانی علاقوں میں بارش کم ہے"),
         ("urea ki bori kitne ki hai?", "یوریا کی بوری کتنے کی ہے؟"), ("spray subah karein ya sham ko?", "سپرے صبح کریں یا شام کو؟"),
         ("dhaan ke baad kya lagayein?", "دھان کے بعد کیا لگائیں؟"), ("deemak se kaise bachein?", "دیمک سے کیسے بچیں؟")]
SYS = ("Convert Roman Urdu into Urdu script. Keep the same words and order. Write English words in Urdu letters. "
       "Output one line of Urdu script only: no notes, no corrections, no other language.")
URDU = re.compile(r"^[؀-ۿݐ-ݿﭐ-﷿ﹰ-\uFEFF0-9\s\-،؟?!.,:()/%+]+$")
def convert(text, strict=False):
    msgs = [{"role": "system", "content": SYS + (" Use only Urdu letters and digits." if strict else "")}]
    for a, b in SHOTS:
        msgs += [{"role": "user", "content": a}, {"role": "assistant", "content": b}]
    x = tokzr.apply_chat_template(msgs + [{"role": "user", "content": text}], add_generation_prompt=True,
                                  return_tensors="pt", return_dict=True).to(llm.device)
    out = tokzr.decode(llm.generate(**x, max_new_tokens=60, do_sample=False)[0][x["input_ids"].shape[1]:], skip_special_tokens=True)
    return re.sub(r"^(Corrected|Note|Urdu)\s*:\s*", "", (out.strip().splitlines() or [""])[0], flags=re.I).strip()
translit, bad = {}, 0
for r in [x for x in qa if x["lang"] == "ur-Latn"]:
    t = convert(r["question"])
    if not URDU.match(t):
        t = convert(r["question"], strict=True)
        bad += not URDU.match(t)
    translit[r["id"]] = t
print(f"conversions {len(translit)}, still not pure Urdu {bad}")
del llm; gc.collect(); torch.cuda.empty_cache()

# 2. Retrievers
FOLD = str.maketrans({"ي": "ی", "ى": "ی", "ك": "ک", "ە": "ہ", "ۀ": "ۂ"})
tok = lambda s: re.findall(r"\w+", unicodedata.normalize("NFKC", s).translate(FOLD).lower())
bm = BM25Okapi([tok(c["text"]) for c in chunks])
enc = SentenceTransformer("BAAI/bge-m3", device="cuda"); enc.max_seq_length = 1024
rer = CrossEncoder("BAAI/bge-reranker-v2-m3", device="cuda", max_length=1024)
C = enc.encode([c["text"] for c in chunks], batch_size=8, normalize_embeddings=True)
qtexts = sorted({r["question"] for r in qa} | set(translit.values()))
E = dict(zip(qtexts, enc.encode(qtexts, batch_size=32, normalize_embeddings=True)))
def rrf(*rows, k=60):
    f = np.zeros(N)
    for s in rows:
        for p, j in enumerate(np.argsort(-s)):
            f[j] += 1 / (k + p + 1)
    return f
def rerank(q, order):
    sc = rer.predict([(q, chunks[j]["text"]) for j in order[:20]], batch_size=8)
    return [order[i] for i in np.argsort(-sc)] + order[20:]
order = lambda s: [int(j) for j in np.argsort(-s)]
def query(r):
    return translit[r["id"]] if r["lang"] == "ur-Latn" else r["question"]
METHODS = {
    "BM25": lambda r: order(bm.get_scores(tok(r["question"]))),
    "BGE-M3": lambda r: order(C @ E[r["question"]]),
    "Ours (routed + conversion + reranker)": lambda r: rerank(query(r), order(C @ E[r["question"]]) if r["lang"] == "en"
                                                              else order(rrf(bm.get_scores(tok(query(r))), C @ E[query(r)]))),
}

# 3. Score on the unseen half, by language and by source
def score(rows, ranks):
    r1, r5, mrr = [], [], []
    for r in rows:
        rk, gold = ranks[r["id"]], set(r["kb_chunk_ids"])
        r1.append(ids[rk[0]] in gold); r5.append(any(ids[j] in gold for j in rk[:5]))
        mrr.append(next((1 / (p + 1) for p, j in enumerate(rk) if ids[j] in gold), 0.0))
    return {"n": len(rows), "r1": round(float(np.mean(r1)), 3), "r5": round(float(np.mean(r5)), 3), "mrr": round(float(np.mean(mrr)), 3)}
test = [r for r in qa if fact(r) in test_facts]
result = {}
for m, fn in METHODS.items():
    ranks = {r["id"]: fn(r) for r in test}
    result[m] = {lang: score([r for r in test if r["lang"] == lang], ranks) for lang in ["ur", "ur-Latn", "en"]}
    print(f"{m:40} " + "  ".join(f"{l} R@5 {v['r5']}" for l, v in result[m].items()))
by_src = defaultdict(list)
for r in test:
    by_src[src[r["kb_chunk_ids"][0]]].append(r)
ranks = {r["id"]: METHODS["Ours (routed + conversion + reranker)"](r) for r in test}
result["ours_by_source"] = {s: score(rows, ranks) for s, rows in sorted(by_src.items())}
print("\nOurs, R@5 by source (all languages):")
for s, v in result["ours_by_source"].items():
    print(f"  {s}  n={v['n']:3}  R@5 {v['r5']}")

json.dump({"kb": "v3", "chunks": N, "questions": len(qa), "split": "facts shuffled seed 42, second half = test",
           "conversion_not_pure_urdu": bad, "results": result, "transliterations": translit},
          open("/kaggle/working/e3_kb_v3.json", "w", encoding="utf-8"), ensure_ascii=False, indent=1)
print("saved /kaggle/working/e3_kb_v3.json")
```
